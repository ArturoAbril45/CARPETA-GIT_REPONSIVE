```python
# Crea tres variables con tu nombre, edad y ciudad
# Muéstralas en pantalla con print()


nombre = "Mario"     # str # float # int
edad = 19
ciudad = "Milagro"
print("Mi nombre es", nombre, "y tengo", edad, "años", "y vivo en", ciudad)
```

```python
# Crea un programa que pida dos números al usuario
# y muestre la suma, resta, multiplicación y división

# input es que pide los datos al usuario 
num1 = float(input("Ingresa un numero: ")) 
num2 = float(input("Ingresa otro numero: "))
suma = num1 + num2
resta = num1 - num2
multiplicacion = num1 * num2
division = num1 / num2
print("La suma es:", suma)           
print("La resta es:", resta)
print("La multiplicación es:", multiplicacion)
print("La división es:", division)
```

```python
# Pide el año de nacimiento del usuario
# Calcula y muestra su edad actual (usa el año 2026)

año_nacimiento = int(input("Ingresa tu año de nacimiento: "))
año_actual = 2026
edad = año_actual - año_nacimiento
print("Tu edad actual es:", edad)
```
