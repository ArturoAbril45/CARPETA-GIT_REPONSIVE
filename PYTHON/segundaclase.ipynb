```python
import pandas as pd

# 1. Crear un DataFrame vacío con las columnas especificadas
df = pd.DataFrame(columns=['Producto', 'Precio', 'Cantidad', 'Categoría'])

# 2. Agregar 5 productos
df.loc[0] = ['Laptop', 850, 10, 'Electrónica']
df.loc[1] = ['Mouse', 25, 50, 'Accesorios']
df.loc[2] = ['Teclado', 45, 30, 'Accesorios']
df.loc[3] = ['Monitor', 200, 15, 'Electrónica']
df.loc[4] = ['Auriculares', 60, 25, 'Accesorios']

print("DataFrame inicial:")
print(df)
print("\n" + "="*50 + "\n")

# 3. Calcular el valor total del inventario de cada producto
df['ValorTotal'] = df['Precio'] * df['Cantidad']

print("DataFrame con ValorTotal:")
print(df)
print("\n" + "="*50 + "\n")

# 4. Mostrar productos con precio mayor a 50
productos_caros = df[df['Precio'] > 50]
print("Productos con precio mayor a 50:")
print(productos_caros)
print("\n" + "="*50 + "\n")

# 5. Encontrar el producto con mayor valor total
producto_mayor_valor = df.loc[df['ValorTotal'].idxmax()]
print("Producto con mayor valor total en inventario:")
print(producto_mayor_valor)
print("\n" + "="*50 + "\n")

# 6. Calcular el total invertido en inventario
total_inventario = df['ValorTotal'].sum()
print(f"Total invertido en inventario: ${total_inventario}")
print("\n" + "="*50 + "\n")

# 7. Mostrar productos ordenados por precio de mayor a menor
df_ordenado = df.sort_values(by='Precio', ascending=False)
print("Productos ordenados por precio (mayor a menor):")
print(df_ordenado)
```

Introducción a Bases de Datos con SQLite en Python 

```python
import sqlite3
```

```python
import sqlite3

conn = sqlite3.connect('escuela.db')
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS estudiantes (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nombre TEXT NOT NULL,
        edad INTEGER,
        grado TEXT,
        calificacion REAL
    )
''')

estudiantes = [
    ('Ana', 14, '8vo', 88.5),
    ('Luis', 15, '9no', 92.0),
    ('Marta', 13, '7mo', 78.0),
    ('Carlos', 16, '10mo', 95.5)
]

cursor.executemany('''
    INSERT INTO estudiantes (nombre, edad, grado, calificacion)
    VALUES (?, ?, ?, ?)
''', estudiantes)

print("TODOS LOS ESTUDIANTES:")
cursor.execute("SELECT * FROM estudiantes")

for fila in cursor.fetchall():
    print(fila)

print("\n" + "="*50 + "\n")

print("ESTUDIANTES CON CALIFICACIÓN > 85:")
cursor.execute('SELECT nombre, calificacion FROM estudiantes WHERE calificacion > 85')

for fila in cursor.fetchall():
    print(f"{fila[0]}: {fila[1]}")

print("\n" + "="*50 + "\n")

cursor.execute('SELECT AVG(calificacion) FROM estudiantes')
promedio = cursor.fetchone()[0]
print(f"Promedio general: {promedio:.2f}")

conn.commit()
conn.close()

print("\nBase de datos 'escuela.db' creada exitosamente")
```

Decoradores en Python

```python
#Crea un decorador llamado anunciar que imprima "INICIO" antes de ejecutar una función 
# "FIN" después de ejecutarla. Aplícalo a una función que salude.
```

```python
def anunciar(func):
    def wrapper():
        print("INICIO")
        func()
        print("FIN")
    return wrapper

@anunciar  # DECORADOR APLICADO A LA FUNCION SALUDAR
def saludar():   #Recursividad es cuando una funcion se llama a si misma
    print("¡Hola! ¿Cómo estás?")

saludar()
```

```python
#Crea un decorador llamado medir_tiempo que calcule y muestre cuánto tiempo tarda en ejecutarse una función. 
#Aplícalo a una función que cuente hasta 1 millón.
```

```python
import time  # Libreria para medir el tiempo de ejecucion

def medir_tiempo(func):
    def wrapper():
        inicio = time.time() # Tiempo de inicio
        func()
        fin = time.time()    # Tiempo de fin
        tiempo_total = fin - inicio
        print(f"Tiempo de ejecución: {tiempo_total:.4f} segundos")
    return wrapper 

@medir_tiempo
def contador():
    total = 0
    for i in range(1, 1000001):
        total += i
    print(f"La suma total es: {total}")
    
contador()
```
